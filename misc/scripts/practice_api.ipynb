```python
import requests
import pandas as pd

url = "https://fantasy.premierleague.com/api/bootstrap-static/"

response = requests.get(url, timeout=30)
response.raise_for_status()
data = response.json()

players = pd.DataFrame(data["elements"])
teams = pd.DataFrame(data["teams"])
gameweeks = pd.DataFrame(data["events"])

display(players.head(20))
print(players.columns.tolist())
```

```python
players = pd.DataFrame(data["elements"])
positions = pd.DataFrame(data["element_types"])

display(positions[["id", "singular_name", "singular_name_short"]])
```

```python
position_map = positions.set_index("id")["singular_name_short"]

players["position"] = players["element_type"].map(position_map)

display(
    players[
        ["id", "first_name", "second_name", "web_name", "position", "form"]
    ].head(20)
)
```

```python
player_names = players[
    ["id", 'code', "first_name", "second_name", "web_name", "expected_goals", "position"]
]

display(player_names.head(20))
```

```python
players["full_name"] = (
    players["first_name"].str.strip()
    + " "
    + players["second_name"].str.strip()
)

display(players[["id", "full_name", "web_name"]].head())
```
